```python
import requests
import torch
import nltk
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer

nltk.download("punkt_tab", quiet=True)

BASE = "http://172.19.0.2:8005"
checkpoint = "/home/lisa/Arupreza/AccentFlow-0.2/FineTuningModels/checkpoints/checkpoint-21500"

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
tokenizer = AutoTokenizer.from_pretrained(checkpoint)
model = AutoModelForSeq2SeqLM.from_pretrained(checkpoint, torch_dtype=torch.bfloat16)
model = model.to(device)
model.eval()

def correct_grammar(text: str) -> str:
    inputs = tokenizer("grammar: " + text, return_tensors="pt", max_length=128, truncation=True).to(device)
    with torch.no_grad():
        outputs = model.generate(**inputs, max_length=128, num_beams=4, early_stopping=True)
    return tokenizer.decode(outputs[0], skip_special_tokens=True)

paragraph = "Hello, I am Emre Rajan Islam from Bangladesh. I am a data science engineer and pursuing my PhD in Sunchung Iyang University. I started my research in 2021 as a Master's Student and my research started with Big Data Analysis on Invehicle Network. What is Invehicle Network? Invehicle Network is inside vehicle called Communication Protocol. In this picture, what we can see inside vehicle, there is a lots of small blocks we can see, which is a small computer, and they need to communication with each other. For that reason, scientists built one communication protocol dedicated for IVN. So it's dedicated for IVN, which is in-vehicle network, and this is called CAN. After introducing CAN, lots of problems solved but few disadvantages also come out. One of them is in CAN there is no inbuilt security mechanism. For that reason, attacker can easily control the car and hack the car. So in this video we can see, see they are controlling the car remotely and the driver don't have any control of his car. All are all are taken by the attacker, controlled by the attacker. So for this reason, the security concern become high on in-vehicle attack. For that reason, scientists start lots of security protocol like intrusion prevention system, intrusion detection system and cryptographical research. But in intuition prevention system there is a big problems come out. Its internal work also normal work also affected by intuition prevention system. So as a solution, ideas is the best solution, but there is several types of ideas introduced, but another drawback comes out with respect to time. Each company, each model, car data is different with each other. For that reason, each car needs individual ideas which is very costly and time-effective, time-consuming. For that reason, our lab and me try to solve this issue that builds one intrusion detection system which will work for all types of vehicles which we can call the universal ideas. So we already developed a few ideas but there is several types of ideas. One is rule-based ideas. Rule-based ideas is like a data analysis and extract some rules from statistical feature or correlation feature after that but after that induce the rules but this type of ideas have lots of false positive rate drawn prediction rate for that reason recently we are applying deep learning using AI with the big data analysis. So in big data analysis we do lots of analysis. We work on several types of attack like DOS, Fuzz, Replay and Stealthy attack. Also we generate data from real car and we analyze the data, we find out the features in normal situation, in attack situation and each component we analyze according to this, we train the model and secure our car as with introducing intr intuition detection system. So till now we already achieved a good result. We make almost universal IDIS prototype. So later on we can see its success. Thank you. Thank you for listening my words"

sentences = nltk.sent_tokenize(paragraph)
print(f"Total sentences: {len(sentences)}\n{'='*80}")

for i, sent in enumerate(sentences, 1):
    corrected = correct_grammar(sent)
    r = requests.post(f"{BASE}/check", json={"text": corrected})
    print(f"[{i:02d}] Input:         {sent}")
    print(f"     Corrected:     {corrected}")
    if r.status_code == 200:
        result = r.json()
        print(f"     Grammar Score: {result['grammar_score']}")
        print(f"     Acceptable:    {result['is_acceptable']}")
    else:
        print(f"     API Error:     {r.status_code} {r.text}")
    print()
```
